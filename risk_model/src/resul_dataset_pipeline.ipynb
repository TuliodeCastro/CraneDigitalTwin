```python
import pandas as pd
import numpy as np

df = pd.read_csv("dataset_original.csv")
```

```python
# mantém só o necessário
# colunas de interesse
cols = [
    "Wind Speed (m/sec)_z1", "Wind Direction (°)_z1",
    "Wind Speed (m/sec)_z2", "Wind Direction (°)_z2",
    "Wind Speed (m/sec)_z3", "Wind Direction (°)_z3",
]
df = df[cols].copy()

df = df.rename(columns={
    "Wind Speed (m/sec)_z1": "wind_speed_z1",
    "Wind Direction (°)_z1": "wind_dir_z1",
    "Wind Speed (m/sec)_z2": "wind_speed_z2",
    "Wind Direction (°)_z2": "wind_dir_z2",
    "Wind Speed (m/sec)_z3": "wind_speed_z3",
    "Wind Direction (°)_z3": "wind_dir_z3",
})

df = df[[
    "wind_speed_z1", "wind_dir_z1",
    "wind_speed_z2", "wind_dir_z2",
    "wind_speed_z3", "wind_dir_z3",
]]
```

```python
for i in (1, 2, 3):
    # converte graus -> radianos (meteorológico -> matemático)
    theta = np.deg2rad(270 - df[f"wind_dir_z{i}"])

    df[f"u{i}"] = df[f"wind_speed_z{i}"] * np.cos(theta)
    df[f"v{i}"] = df[f"wind_speed_z{i}"] * np.sin(theta)
```

```python
# baricentro (pesos iguais)
df["u_c"] = (df["u1"] + df["u2"] + df["u3"]) / 3
df["v_c"] = (df["v1"] + df["v2"] + df["v3"]) / 3
```

```python
# magnitude e direção
df["result_vel"] = np.sqrt(df["u_c"]**2 + df["v_c"]**2)
df["result_dir"] = np.degrees(np.arctan2(df["v_c"], df["u_c"]))
df["result_dir"] = (df["result_dir"] + 360) % 360
```

```python
# dataframe final
df_final = df[["result_vel", "result_dir"]]
```

```python
# salvar novo csv
df_final.to_csv("result_wind_center.csv", index=False)
```
